***LEY DE FOURIER***

Sea J la densidad de corriente de energía (energía por unidad de área y por unidad de tiempo), que se establece en la barra debido a la diferencia de temperaturas entre dos puntos de la misma. La ley de Fourier afirma que hay una proporcionalidad entre el flujo de energía J y el gradiente de temperatura.

$$J=-k\frac{\partial T}{\partial x}$$

Consideremos un elemento de la barra de longitud $dx$ y sección transversal $S$. La energía que entra en el elemento de volumen por unidad de tiempo es $JS$, mientras que la que sale es $J'S$. La variación de energía del elemento por unidad de tiempo es igual a la diferencia entre el flujo entrante y el flujo saliente:

$$
JS - J'S = -\frac{\partial J}{\partial x}S\,dx.
$$

Esta energía se emplea en modificar la temperatura del elemento. La cantidad de energía absorbida o cedida por unidad de tiempo es igual al producto de la masa del elemento, su calor específico y la variación temporal de la temperatura:

$$
(\rho S\,dx)c\frac{\partial T}{\partial t}.
$$

Igualando ambas expresiones y utilizando la ley de Fourier,

$$
J = -K\frac{\partial T}{\partial x},
$$

se obtiene la ecuación diferencial que describe la conducción térmica del calor:

$$
\frac{\partial T}{\partial t}
=
\alpha
\frac{\partial^2 T}{\partial x^2},
$$

donde la difusividad térmica $\alpha$ está definida por

$$
\alpha = \frac{K}{\rho c}.
$$

***Solución analítica***

Supongamos una barra metálica de longitud $L$, conectada por sus extremos a dos focos de calor a temperaturas $T_a$ y $T_b$, respectivamente. Sea $T_0$ la temperatura inicial de la barra cuando se conectan los focos a sus extremos.

Al cabo de cierto tiempo, teóricamente infinito (aunque en la práctica depende del material empleado), se establece un estado estacionario en el que la temperatura de cada punto de la barra no varía con el tiempo. Dicho estado está caracterizado por un flujo de energía constante $J$. La ley de Fourier establece que la temperatura variará linealmente con la distancia $x$ al origen de la barra:

$$
T(x)=T_a+\frac{T_b-T_a}{L}x
$$

Para describir el estado transitorio buscamos una solución de la forma

$$
T(x,t)=F(x)G(t),
$$

es decir, una solución por separación de variables.

La ecuación de difusión del calor es

$$
\frac{\partial T(x,t)}{\partial t}
=
\alpha
\frac{\partial^2 T(x,t)}{\partial x^2},
$$

donde $\alpha$ es la difusividad térmica.

Sustituyendo la solución separable obtenemos

$$
\frac{1}{\alpha}
\frac{1}{G(t)}
\frac{dG(t)}{dt}
=
\frac{1}{F(x)}
\frac{d^2F(x)}{dx^2}
=
-\omega^2.
$$

El signo negativo asegura el carácter transitorio de la solución.

\subsection{Dependencia temporal}

La ecuación diferencial para $G(t)$ es

$$
\frac{dG(t)}{dt}
+
\alpha \omega^2 G(t)
=
0.
$$

Su solución es

$$
G(t)=G(0)e^{-\alpha \omega^2 t}.
$$

\subsection{Dependencia espacial}

La ecuación diferencial para $F(x)$ es

$$
\frac{d^2F(x)}{dx^2}
+
\omega^2F(x)
=
0.
$$

Esta ecuación es análoga a la del movimiento armónico simple, cuya solución general es

$$
F(x)=A\sin(\omega x+\delta).
$$

Por tanto, la temperatura en cualquier punto de la barra y en cualquier instante puede expresarse como la suma de una solución estacionaria y una solución transitoria:

$$
T(x,t)
=
T_a+\frac{T_b-T_a}{L}x
+
\sum_{n=1}^{\infty}
a_n
e^{-\alpha \omega_n^2 t}
\sin(\omega_n x+\delta_n).
$$

\subsection{Condiciones de contorno}

En el extremo izquierdo de la barra:

$$
T(0,t)=T_a.
$$

Entonces

$$
0=
\sum_{n=1}^{\infty}
a_n
e^{-\alpha \omega_n^2 t}
\sin(\delta_n),
$$

lo que implica

$$
\delta_n=0.
$$

En el extremo derecho:

$$
T(L,t)=T_b.
$$

Por tanto,

$$
0=
\sum_{n=1}^{\infty}
a_n
e^{-\alpha \omega_n^2 t}
\sin(\omega_nL),
$$

de donde se obtiene

$$
\omega_nL=n\pi,
$$

y por consiguiente

$$
\omega_n=\frac{n\pi}{L}.
$$

La solución general queda

$$
T(x,t)
=
T_a+\frac{T_b-T_a}{L}x
+
\sum_{n=1}^{\infty}
a_n
\exp\left(
-\alpha\frac{n^2\pi^2}{L^2}t
\right)
\sin\left(
\frac{n\pi x}{L}
\right).
$$

\subsection{Distribución inicial de temperaturas}

Para determinar los coeficientes $a_n$, imponemos la condición inicial

$$
T(x,0)=T_0.
$$

Entonces

$$
T_0
=
T_a+\frac{T_b-T_a}{L}x
+
\sum_{n=1}^{\infty}
a_n
\sin\left(
\frac{n\pi x}{L}
\right).
$$

Definimos

$$
f(x)
=
T_0
-
T_a
-
\frac{T_b-T_a}{L}x.
$$

Por tanto,

$$
f(x)
=
\sum_{n=1}^{\infty}
a_n
\sin\left(
\frac{n\pi x}{L}
\right).
$$

Los coeficientes $a_n$ se obtienen mediante el desarrollo en serie de Fourier seno:

$$
a_n=
\begin{cases}
\dfrac{2}{n\pi}
\left(
2T_0-T_a-T_b
\right),
& n \text{ impar},
\\[0.4cm]
\dfrac{2}{n\pi}
\left(
T_b-T_a
\right),
& n \text{ par}.
\end{cases}
$$

Finalmente, la distribución de temperatura en cualquier instante queda dada por

$$
T(x,t)
=
T_a+\frac{T_b-T_a}{L}x
+
\sum_{n=1}^{\infty}
a_n
\exp\left(
-\alpha\frac{n^2\pi^2}{L^2}t
\right)
\sin\left(
\frac{n\pi x}{L}
\right).
$$

El parámetro

$$
\alpha=\frac{K}{\rho c}
$$

es la difusividad térmica del material, donde:

$K$ es la conductividad térmica.
$\rho$ es la densidad.
$c$ es el calor específico.


El valor de $\alpha$ proporciona una medida de la rapidez con la que el sistema alcanza el estado estacionario. Cuanto mayor sea $\alpha$, más rápidamente se estabilizará la distribución de temperatura en la barra.

```python
#librerias utilizadas
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
```

```python
plt.rcParams["animation.html"] = "jshtml"
```

```python
#Propiedades del acero
k =45        # W/(m K)
rho = 7800   # kg/m^3
c =460       # J/(kg K)
alpha=k/(rho*c)
```

```python
# Caracterización de la barra
L = 1.0                 # longitud (m)
N = 50                  # nodos
dx = L/(N-1)

# vector espacial
x = np.linspace(0, L, N)

#Condiciones de frontera
Ta = 100                # extremo izquierdo
Tb = 20                  # extremo derecho

#temperatura inicial
T0 = np.zeros(N)
T0[0] = Ta
T0[-1] = Tb
```

```python
#Sistema de ecuaciones
def f(t, T):

    dTdt = np.zeros_like(T)

    # extremos fijos
    dTdt[0] = 0
    dTdt[-1] = 0

    # nodos interiores
    for i in range(1, N-1):
        dTdt[i] = alpha*(T[i+1]-2*T[i]+T[i-1])/dx**2

    return dTdt

#Euler
def euler(f, x0, y0, x1, n):
    h = (x1 - x0)/n
    xi= np.zeros(n+1)
    yi= np.zeros((n+1, len(y0)))

    xi[0] = x0
    yi[0] = y0

    for i in range(n):

        k1 = f(xi[i], yi[i])

        yi[i+1] = yi[i] + h*k1

        yi[i+1,0] = Ta
        yi[i+1,-1] = Tb

        xi[i+1] = xi[i] + h

    return xi, yi

# Runge-Kutta 2
def rungekutta2(f,x0,y0,x1,n):

    h =(x1-x0)/n
    xi = np.zeros(n+1)
    yi = np.zeros((n+1, len(y0)))

    xi[0] = x0
    yi[0] = y0

    for i in range(n):

        k1 = f(xi[i], yi[i])
        k2 = f(xi[i]+h/2, yi[i]+h*k1/2)

        yi[i+1] = yi[i] + h*k2

        yi[i+1,0] = Ta
        yi[i+1,-1] = Tb

        xi[i+1] = xi[i] + h

    return xi, yi


#Runge-Kutta 4
def rungekutta4(f, x0, y0, x1, n):

    h = (x1-x0)/n

    xi = np.zeros(n+1)
    yi = np.zeros((n+1, len(y0)))

    xi[0] = x0
    yi[0] = y0

    for i in range(n):

        k1 = f(xi[i], yi[i])
        k2 = f(xi[i]+h/2, yi[i]+h*k1/2)
        k3 = f(xi[i]+h/2, yi[i]+h*k2/2)
        k4 = f(xi[i]+h, yi[i]+h*k3)

        yi[i+1] = yi[i] + h*(k1+2*k2+2*k3+k4)/6

        yi[i+1,0] = Ta
        yi[i+1,-1] = Tb

        xi[i+1] = xi[i] + h

    return xi, yi


# SIMULACION
t0 = 0
tf = 1000
n = 1000

te, Ye = euler(f, t0, T0, tf, n)
trk2, Yrk2 = rungekutta2(f, t0, T0, tf, n)
trk4, Yrk4 = rungekutta4(f, t0, T0, tf, n)

#Graficas
#PERFIL FINAL DE TEMPERATURA
fig, ax = plt.subplots(1, 3, figsize=(15,5))

ax[0].plot(x, Ye[-1], 'b')
ax[0].set_title("Euler")
ax[0].set_xlabel("Posición x (m)")
ax[0].set_ylabel("Temperatura (°C)")
ax[0].grid(True)

ax[1].plot(x, Yrk2[-1], 'g')
ax[1].set_title("RK2")
ax[1].set_xlabel("Posición x (m)")
ax[1].grid(True)

ax[2].plot(x, Yrk4[-1], 'r')
ax[2].set_title("RK4")
ax[2].set_xlabel("Posición x (m)")
ax[2].grid(True)

plt.tight_layout()
plt.show()

# EVOLUCION DEL NODO CENTRAL

nodo = N // 2

plt.figure(figsize=(8,5))

plt.plot(te, Ye[:, nodo], label='Euler')
plt.plot(trk2, Yrk2[:, nodo], label='RK2')
plt.plot(trk4, Yrk4[:, nodo], label='RK4')

plt.xlabel("Tiempo (s)")
plt.ylabel("Temperatura (°C)")
plt.title("Temperatura en el nodo central")
plt.legend()
plt.grid(True)

plt.show()

# COMPROBACION DE ESTABILIDAD

dt = (tf - t0)/n
r = alpha*dt/dx**2

print("alpha =", alpha)
print("dx =", dx)
print("dt =", dt)
print("r =", r)
```

```python
# FIGURA
fig, ax = plt.subplots(3, 1, figsize=(12, 5))

vmin = min(np.min(Ye), np.min(Yrk2), np.min(Yrk4))
vmax = max(np.max(Ye), np.max(Yrk2), np.max(Yrk4))


# EULER
im1 = ax[0].imshow(
    Ye[0].reshape(1, -1),
    cmap='viridis',
    aspect='auto',
    extent=[0, L, 0, 0.1],
    vmin=vmin,
    vmax=vmax
)

ax[0].set_yticks([])
ax[0].set_title("Euler")

# RK2
im2 = ax[1].imshow(
    Yrk2[0].reshape(1, -1),
    cmap='viridis',
    aspect='auto',
    extent=[0, L, 0, 0.1],
    vmin=vmin,
    vmax=vmax
)

ax[1].set_yticks([])
ax[1].set_title("Runge-Kutta 2")

# RK4
im3 = ax[2].imshow(
    Yrk4[0].reshape(1, -1),
    cmap='viridis',
    aspect='auto',
    extent=[0, L, 0, 0.1],
    vmin=vmin,
    vmax=vmax
)

ax[2].set_yticks([])
ax[2].set_title("Runge-Kutta 4")
ax[2].set_xlabel("Posición x (m)")


# BARRA DE COLOR
cbar = fig.colorbar(
    im3,
    ax=ax,
    orientation='vertical',
    pad=0.02
)

cbar.set_label("Temperatura (°C)")


# TEXTO DE TIEMPO
time_text = fig.suptitle(
    "t = 0.0 s",
    fontsize=14
)

#ACTUALIZACION
def update(frame):

    im1.set_array(
        Ye[frame].reshape(1,-1)
    )

    im2.set_array(
        Yrk2[frame].reshape(1,-1)
    )

    im3.set_array(
        Yrk4[frame].reshape(1,-1)
    )

    time_text.set_text(
        f"t = {te[frame]:.1f} s"
    )

    return im1, im2, im3

frames = range(0, len(te), 10)

# ANIMACION

ani = FuncAnimation(
    fig,
    update,
    frames=frames,
    interval=50,
    blit=False
)

HTML(ani.to_jshtml())
```
